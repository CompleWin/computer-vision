```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

```python
def plot_lut(lut_table: list):
    plt.figure(figsize=(4, 4))
    plt.plot(list(range(len(lut_table))), lut_table)
    plt.show()
```

```python
img_path = "/Users/alexey/Downloads/test_image_2.jpeg"

img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)

plt.figure(figsize=(8, 6))
plt.imshow(img, "gray")
plt.axis("off")
plt.show()
```

```python
lut_table = np.arange(0, 256, dtype=np.uint8)

for i in range(len(lut_table)):
    lut_table[i] = int(255 * np.pow(lut_table[i] / 255.0, 0.8))

# new_img = img.copy()
# for i in range(img.shape[0]):
#     for j in range(img.shape[1]):
#         new_img[i, j] = lut_table[img[i, j]]

new_img = cv2.LUT(img, lut_table)

plot_lut(lut_table)

plt.figure(figsize=(8, 6))
plt.imshow(new_img, "gray")
plt.axis("off")
plt.show()
```
